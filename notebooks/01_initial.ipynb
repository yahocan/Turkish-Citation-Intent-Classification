```python
import kagglehub

kagglehub.auth.login(validate_credentials=False)
```

```python
from pathlib import Path
import shutil

path = kagglehub.competition_download("citation-intent-classification-competitions")

data_dir = Path("data")
data_dir.mkdir(exist_ok=True)

for file in Path(path).iterdir():
    if file.is_file():
        shutil.copy2(file, data_dir / file.name)

print("Project data folder:", data_dir.resolve())
print("Files:", sorted(p.name for p in data_dir.iterdir()))
```
